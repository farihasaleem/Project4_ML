# Splitting train test 

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop(columns=['diagnosis'])
y=df['diagnosis']

x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.8,random_state=42,stratify=y)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
```

```python
from xgboost import XGBClassifier
```

# Model Selection

```python
models={'LogisticRegression': Pipeline([('scalar', StandardScaler()),
                                       ('clf',LogisticRegression(max_iter=1000))]),
       'SVM':Pipeline([('scalar',StandardScaler()),
                      ('clf',SVC(probability=True))]),
       'RandomForest':Pipeline([('clf',RandomForestClassifier(random_state=42))]),
       'XGBoost':Pipeline([('clf',XGBClassifier(eval_metric='logloss',random_state=42))])}
```

# Comparing Models

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
```

```python
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
result=[]
```

```python
for name, model in models.items():
    scores= cross_validate(model, x_train,y_train, cv=cv, 
                          scoring=['accuracy','precision','recall','roc_auc'])
    result.append({
        'Model':name,
        'Accuracy':scores['test_accuracy'].mean(),
        'Precision':scores['test_precision'].mean(),
        'Recall':scores['test_recall'].mean(),
        'ROC-AUC':scores['test_roc_auc'].mean()
    })
```

```python
pd.DataFrame(result).sort_values('Recall',ascending=False)
```

# Tuning best 1-2 Models

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid={'clf__C':[0.01,0.1,1,10,100]}

grid=GridSearchCV(models['LogisticRegression'], param_grid,cv=cv,scoring='recall')

grid.fit(x_train,y_train)

print(grid.best_params_)
```

```python
print(grid.best_score_)
```

# Evaluation

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, ConfusionMatrixDisplay, RocCurveDisplay
```

```python
best_model=grid.best_estimator_
```

```python
y_pred=best_model.predict(x_test)

y_prob=best_model.predict_proba(x_test)[:,1]

print(classification_report(y_test,y_pred, target_names=['Benign','Malignant']))
print('ROC-AUC ', roc_auc_score(y_test,y_prob))

ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
RocCurveDisplay.from_predictions(y_test,y_prob)

plt.show()
```
